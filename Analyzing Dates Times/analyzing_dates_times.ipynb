### Demo

```python
import pandas as pd

# Load the dataset with the 'Date' column parsed as datetime
crypto = pd.read_csv('cryptocurrency_BTC.csv', parse_dates=['Date'])

# Define the start and end dates for the analysis
start_date = pd.to_datetime('2023-05-01')
end_date = pd.to_datetime('2024-05-01')

# Use query to select the close prices for the start and end dates
close_start = crypto.query('Date == @start_date')['Close'].iloc[0]
close_end = crypto.query('Date == @end_date')['Close'].iloc[0]


# Calculate percentage growth from start to end date
growth = ((close_end - close_start) / close_start) * 100

# Print the results
print(growth)
```
