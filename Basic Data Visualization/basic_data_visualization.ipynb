## Bar Charts

```python
transactions.groupby(['store_nbr']).agg({'transactions': 'sum'}).plot.bar(figsize=(10,10))
```

```python
store_40s = list(range(40,50))
```

```python
(transactions
     .query('store_nbr in @store_40s')
     .groupby(['store_nbr'])
     .agg({'transactions': 'sum'})
     .plot.bar(figsize=(10,10))
)
```

```python
(transactions
     .query('store_nbr in @store_40s')
     .groupby(['store_nbr'])
     .agg({'transactions': 'sum'})
     .sort_values(by='transactions')
     .plot.bar(figsize=(10,10))
)
```

```python
(transactions
     .query('store_nbr in @store_40s')
     .groupby(['store_nbr'])
     .agg({'transactions': 'sum'})
     .sort_values(by='transactions')
     .plot.barh(figsize=(10,10))
)
```

## Grouped & Stacked Bar Charts

```python
(transactions
     .query('store_nbr in @store_40s and date.dt.month in [1, 2, 3]')
     .pivot_table(
         index=transactions['date'].dt.month,
         columns=transactions['store_nbr'],
         values='transactions',
         aggfunc='sum'
     )
)
```

```python
(transactions
     .query('store_nbr in @store_40s and date.dt.month in [1, 2, 3]')
     .pivot_table(
         index=transactions['date'].dt.month,
         columns=transactions['store_nbr'],
         values='transactions',
         aggfunc='sum'
     )
).plot.bar().legend(bbox_to_anchor=(1,1))
```

```python
(transactions
     .query('store_nbr in @store_40s and date.dt.month in [1, 2, 3]')
     .pivot_table(
         index=transactions['date'].dt.month,
         columns=transactions['store_nbr'],
         values='transactions',
         aggfunc='sum'
     )
).plot.bar(stacked=True).legend(bbox_to_anchor=(1,1))
```

```python
(transactions
     .query('store_nbr in @store_40s and date.dt.month in [1, 2, 3]')
     .pivot_table(
         index=transactions['date'].dt.month,
         columns=transactions['store_nbr'],
         values='transactions',
         aggfunc='sum'
     )
).plot.barh(stacked=True).legend(bbox_to_anchor=(1,1))
```

```python
(transactions
     .query('store_nbr in @store_40s and date.dt.month in [1, 2, 3]')
     .pivot_table(
         index=transactions['date'].dt.month,
         columns=transactions['store_nbr'],
         values='transactions',
         aggfunc='sum'
     ).apply(lambda x: x * 100 / sum(x), axis = 1)
).plot.bar(stacked=True).legend(bbox_to_anchor=(1,1))
```
