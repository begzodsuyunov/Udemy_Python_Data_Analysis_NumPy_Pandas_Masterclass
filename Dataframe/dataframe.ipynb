## Advanced Conditional Columns with Select

```python
oil = pd.read_csv(r"D:\Self_Study\Udemy\Python\Python Data Analysis NumPy & Pandas Masterclass\Pandas+Course+Resources\Pandas Course Resources\retail\oil.csv")
```

```python
oil.columns = ['price_date', 'price']
```

```python
oil
```

```python
import numpy as np

oil['buy'] = np.where(oil['price'] > 100, 'Too High', 'Buy')

oil
```

```python
oil['buy'].value_counts()
```

```python
conditions = [
    (oil['price'] > 100),
    (oil['price'] <= 100) & (oil['price'] > 50),
    (oil['price'] <= 50)
]

choices = ["Don't Buy", "Buy", "Strong Buy"]

oil['buy'] = np.select(conditions, choices, default='Missing')

oil.head()
```

```python
oil['buy'].value_counts()
```
