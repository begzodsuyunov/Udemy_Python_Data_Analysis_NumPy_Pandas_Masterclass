## Agg Method & Names Aggregations

```python
premier_league[['season', 'HomeTeam', 'id', 'HomeGoals', 'AwayGoals']].groupby(['season', 'HomeTeam']).agg(['sum', 'mean'])
```

```python
(premier_league[['season', 'HomeTeam', 'id', 'HomeGoals', 'AwayGoals']]
 .groupby(['season', 'HomeTeam'])
 .agg({'HomeGoals': ['sum', 'mean'], 'AwayGoals': ['sum', 'mean']}))
```

```python
(premier_league[['season', 'HomeTeam', 'id', 'HomeGoals', 'AwayGoals']]
 .groupby(['season', 'HomeTeam'], as_index=False)
 .agg(home_goal_sum = ('HomeGoals', 'sum'),
      away_goal_sum = ('AwayGoals', 'sum')))
```

### Demo

```python
import pandas as pd

# Load the dataset
insurance_df = pd.read_csv('insurance_data.csv')

# Group by 'region' and 'smoker', then calculate multiple aggregations
aggregated_data = insurance_df.groupby(['region', 'smoker']).agg(
    average_charges = ('charges', 'mean'),
    total_charges = ('charges', 'sum'),
    max_children = ('children', 'max')
)

# Display the aggregated data
aggregated_data
```

### Assignment 4

```python
# Recreate table from section 3

transactions2 = transactions.assign(
    target_pct=transactions["transactions"] / 2500,
    met_target=(transactions["transactions"] / 2500) >= 1,
    bonus_payable=((transactions["transactions"] / 2500) >= 1) * 100,
    month=transactions.date.dt.month,
    day_of_week=transactions.date.dt.dayofweek,
)

transactions2.head()
```

```python
(transactions2.groupby('store_nbr')
 .agg(
     met_target = ('met_target', 'mean'),
     bonus_payable = ('bonus_payable', 'sum')
 )
 .sort_values('bonus_payable', ascending=False)
 .head()
)
```

```python
(transactions2.groupby('month')
 .agg(
     met_target = ('met_target', 'mean'),
     bonus_payable = ('bonus_payable', 'sum')
 )
 .sort_values('bonus_payable', ascending=False)
 .head()
)
```

```python
(transactions2.groupby('day_of_week')
 .agg(
     met_target = ('met_target', 'mean'),
     bonus_payable = ('bonus_payable', 'sum')
 )
 .sort_values('bonus_payable', ascending=False)
 .head()
)
```
